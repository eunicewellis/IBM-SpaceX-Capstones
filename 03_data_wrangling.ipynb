# SpaceX Falcon 9 Data Wrangling
Uses IBM Skills Network course datasets when accessible. Outputs are written to `../data`. Class is the course target label; use course label definitions consistently.

```python
import pandas as pd
from pathlib import Path
DATA_DIR = Path('../data')
DATA_DIR.mkdir(exist_ok=True)
base = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/'
```

```python
df = pd.read_csv(base + 'dataset_part_2.csv')
print('shape:', df.shape)
display(df.head())
display(df.isna().sum().sort_values(ascending=False).head(15))
```

```python
# Inspect duplicates, types and target balance
print('duplicate rows:', df.duplicated().sum())
display(df.dtypes)
if 'Class' in df.columns:
    display(df['Class'].value_counts(dropna=False).rename('count').to_frame())
```

```python
# Conservative cleanup: remove exact duplicate rows; keep missing values visible for documentation.
clean = df.drop_duplicates().copy()
print('shape after duplicate removal:', clean.shape)
clean.to_csv(DATA_DIR / 'dataset_part_2_clean.csv', index=False)
clean.head()
```

## Notes on missing values and leakage
Do not impute/encode the entire dataset before a train/test split for a prospective model. The course lab's prepared feature matrix may already include preprocessing. For a stronger evaluation, put imputation, encoding and scaling inside an sklearn Pipeline fitted only on training folds. Keep this cleanup output separate from source data.
