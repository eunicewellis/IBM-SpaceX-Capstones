# SpaceX Falcon 9 Data Collection — Web Scraping
Collects a historical launch table from Wikipedia using pandas' HTML table reader. Website markup and access policies may change; review source attribution before redistribution.

```python
import pandas as pd
from pathlib import Path
DATA_DIR = Path('../data')
DATA_DIR.mkdir(exist_ok=True)
source_url = 'https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches'
```

```python
tables = pd.read_html(source_url)
print('Tables discovered:', len(tables))
[(i, t.shape, list(t.columns)[:5]) for i, t in enumerate(tables[:10])]
```

```python
# Inspect tables to identify the historical launch table before choosing one.
for i, table in enumerate(tables):
    print(i, table.shape, list(map(str, table.columns))[:8])
```

```python
# Save a combined copy of the tables as a transparent extraction artifact.
# This avoids guessing which table is correct if Wikipedia changes its layout.
with pd.ExcelWriter(DATA_DIR / 'wikipedia_launch_tables.xlsx') as writer:
    for i, table in enumerate(tables):
        table.to_excel(writer, sheet_name=f'table_{i}'[:31], index=False)
print('Saved tables to data/wikipedia_launch_tables.xlsx')
```
