## Tasks

Now write and execute SQL queries to solve the assignment tasks.

### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
# Task 1: Display the first five rows
%sql SELECT * FROM SPACEXTABLE LIMIT 5;
```


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA' 


```python
# Task 2: How many launches occurred?
%sql SELECT COUNT(*) AS total_launches FROM SPACEXTABLE;
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
# Task 3: List unique launch sites
%sql SELECT DISTINCT Launch_Site FROM SPACEXTABLE;
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
# Task 4: Count launches for each launch site
%sql SELECT Launch_Site, COUNT(*) AS launches FROM SPACEXTABLE GROUP BY Launch_Site ORDER BY launches DESC;
```

### Task 5

##### List the date when the first succesful landing outcome in ground pad was acheived.


_Hint:Use min function_ 


```python
# Task 5: Find 5 records with total payload mass greater than 10000 kg
%sql SELECT * FROM SPACEXTABLE WHERE PAYLOAD_MASS__KG_ > 10000 LIMIT 5;
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
# Task 6: Average payload mass by booster version
%sql SELECT Booster_Version, AVG(PAYLOAD_MASS__KG_) AS avg_payload_kg FROM SPACEXTABLE GROUP BY Booster_Version ORDER BY avg_payload_kg DESC;
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
# Task 7: Successful landing outcomes (class = 1) by landing outcome
%sql SELECT Landing_Outcome, COUNT(*) AS successful_records FROM SPACEXTABLE WHERE Class = 1 GROUP BY Landing_Outcome ORDER BY successful_records DESC;
```

### Task 8



##### List all the booster_versions that have carried the maximum payload mass, using a subquery with a suitable aggregate function.


```python
# Task 8: Total and successful launches by site
%sql SELECT Launch_Site, COUNT(*) AS total_launches, SUM(Class) AS successful_landings, ROUND(AVG(Class)*100, 2) AS success_rate_pct FROM SPACEXTABLE GROUP BY Launch_Site ORDER BY success_rate_pct DESC;
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
# Task 9: Successful launches for booster versions with 'Block 5' in the name
%sql SELECT Booster_Version, COUNT(*) AS successful_landings FROM SPACEXTABLE WHERE Class = 1 AND Booster_Version LIKE '%Block 5%' GROUP BY Booster_Version;
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order.


```python
# Task 10: Payload statistics for successful launches
%sql SELECT MIN(PAYLOAD_MASS__KG_) AS min_payload_kg, AVG(PAYLOAD_MASS__KG_) AS avg_payload_kg, MAX(PAYLOAD_MASS__KG_) AS max_payload_kg FROM SPACEXTABLE WHERE Class = 1;
```
